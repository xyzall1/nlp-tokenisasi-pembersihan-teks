## 1. Import Library

```python
import re
import string
from collections import Counter

# Install NLTK
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

print('Library berhasil diimport!')
```

## 2. Teks Asli (Raw Text)

```python
teks_asli = """Twenty-first-century politics has witnessed an alarming rise of populism in the U.S. and Europe. The first warning signs came with the UK Brexit Referendum vote in 2016 swinging in the way of Leave. This was followed by a stupendous victory by billionaire Donald Trump to become the 45th President of the United States in November 2016. Since then, Europe has seen a steady rise in populist and far-right parties that have capitalized on Europe's Immigration Crisis to raise nationalist and anti-Europe sentiments. Some instances include Alternative for Germany (AfD) winning 12.6% of all seats and entering the Bundestag, thus upsetting Germany's political order for the first time since the Second World War, the success of the Five Star Movement in Italy and the surge in popularity of neo-nazism and neo-fascism in countries such as Hungary, Czech Republic, Poland and Austria."""

print('TEKS ASLI :')
print(teks_asli)
```

## 3. Tokenisasi Kalimat (Sentence Tokenization)

```python
kalimat_tokens = sent_tokenize(teks_asli)

print("TOKENISASI KALIMAT:")
for i, kalimat in enumerate(kalimat_tokens, 1):
    print(f'Kalimat {i}: {kalimat}')
    print()

print(f'Jumlah kalimat: {len(kalimat_tokens)}')
```

## 4. Tokenisasi Kata (Word Tokenization)

```python
kata_tokens_raw = word_tokenize(teks_asli)

print('TOKENISASI KATA (RAW): ')
print(kata_tokens_raw)
print(f'\nJumlah token (termasuk tanda baca): {len(kata_tokens_raw)}')
```

## 5. Pembersihan Teks (Text Cleaning)

Langkah-langkah pembersihan:
1. **Lowercasing** untuk konversi ke huruf kecil
2. **Remove Numbers** untuk hapus angka
3. **Remove Punctuation** untuk hapus tanda baca
4. **Remove Extra Whitespace** untuk hapus spasi berlebih
5. **Remove Stopwords** untuk hapus kata umum
6. **Stemming** untuk ubah kata ke bentuk dasar (metode Porter)
7. **Lemmatization** untuk ubah kata ke bentuk kamus

```python
# Langkah 1 (Lowercasing)
teks_lower = teks_asli.lower()
print('LANGKAH 1 LOWERCASING')
print(teks_lower)
```

```python
# Langkah 2 (Hapus Angka)
teks_tanpa_angka = re.sub(r'\d+\.?\d*\%?', '', teks_lower)
print('LANGKAH 2 HAPUS ANGKA')
print(teks_tanpa_angka)
```

```python
# Langkah 3 (Hapus Tanda Baca)
teks_tanpa_tanda_baca = re.sub(r'[^\w\s]', ' ', teks_tanpa_angka)
print('LANGKAH 3 HAPUS TANDA BACA')
print(teks_tanpa_tanda_baca)
```

```python
# Langkah 4 (Hapus Whitespace Berlebih)
teks_bersih = re.sub(r'\s+', ' ', teks_tanpa_tanda_baca).strip()
print('LANGKAH 4 HAPUS WHITESPACE BERLEBIH')
print(teks_bersih)
```

```python
# Langkah 5 (Tokenisasi dan Hapus Stopwords)
stop_words = set(stopwords.words('english'))
tokens_bersih = word_tokenize(teks_bersih)
tokens_tanpa_stopwords = [kata for kata in tokens_bersih if kata not in stop_words]

print('LANGKAH 5 HAPUS STOPWORDS')
print(tokens_tanpa_stopwords)
print(f'\nJumlah token sebelum hapus stopwords : {len(tokens_bersih)}')
print(f'Jumlah token setelah hapus stopwords  : {len(tokens_tanpa_stopwords)}')
```

```python
# Langkah 6 (Stemming (Porter Stemmer)
stemmer = PorterStemmer()
tokens_stem = [stemmer.stem(kata) for kata in tokens_tanpa_stopwords]

print('LANGKAH 6 STEMMING')
print(tokens_stem)
print('\n--- Perbandingan Kata Asli vs Hasil Stemming (15 kata pertama) ---')
print(f'{"Kata Asli":<25} --> {"Hasil Stem"}')
print('-' * 45)
for asli, stem in zip(tokens_tanpa_stopwords[:15], tokens_stem[:15]):
    print(f'{asli:<25} --> {stem}')
```

```python
# Langkah 7 (Lemmatization)
lemmatizer = WordNetLemmatizer()
tokens_lemma = [lemmatizer.lemmatize(kata) for kata in tokens_tanpa_stopwords]

print('LANGKAH 7: LEMMATIZATION')
print(tokens_lemma)
print('\n--- Perbandingan Kata Asli vs Hasil Lemmatization (15 kata pertama) ---')
print(f'{"Kata Asli":<25} --> {"Hasil Lemma"}')
print('-' * 45)
for asli, lemma in zip(tokens_tanpa_stopwords[:15], tokens_lemma[:15]):
    print(f'{asli:<25} --> {lemma}')
```

## 6. Analisis Frekuensi Kata (Word Frequency Analysis)

```python
frekuensi = Counter(tokens_tanpa_stopwords)
kata_paling_sering = frekuensi.most_common(10)

print('=== 10 KATA PALING SERING MUNCUL ===')
print(f'{"Kata":<25} {"Frekuensi"}')
print('-' * 35)
for kata, frek in kata_paling_sering:
    print(f'{kata:<25} {frek}')
```

## 7. Ringkasan Hasil

```python
print('       RINGKASAN HASIL PEMROSESAN TEKS')
print('=' * 55)
print(f'Jumlah kalimat                     : {len(kalimat_tokens)}')
print(f'Jumlah token awal (raw)            : {len(kata_tokens_raw)}')
print(f'Jumlah token setelah pembersihan   : {len(tokens_bersih)}')
print(f'Jumlah token tanpa stopwords       : {len(tokens_tanpa_stopwords)}')
print(f'Jumlah kata unik (tanpa stopwords) : {len(set(tokens_tanpa_stopwords))}')
print('=' * 55)
print('\nToken bersih (tanpa stopwords):')
print(' '.join(tokens_tanpa_stopwords))
print('\nToken hasil Stemming:')
print(' '.join(tokens_stem))
print('\nToken hasil Lemmatization:')
print(' '.join(tokens_lemma))
```
